# Fraud Detection Project - Baseline Modeling

## Objective
This notebook trains baseline machine learning models for fraud detection using the cleaned and preprocessed dataset. The goal is to compare simple baseline models and evaluate their ability to identify fraudulent transactions in a highly imbalanced classification setting.

#### Import libraries

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score
```

Load processed datasets

```python
X_train = pd.read_csv(r"C:\Users\Jairus\OneDrive\Documents\Portfolio\01.Fraud_Detection_Project\data\processed\X_train.csv")
X_test = pd.read_csv(r"C:\Users\Jairus\OneDrive\Documents\Portfolio\01.Fraud_Detection_Project\data\processed\X_test.csv")
y_train = pd.read_csv(r"C:\Users\Jairus\OneDrive\Documents\Portfolio\01.Fraud_Detection_Project\data\processed\y_train.csv").squeeze()
y_test = pd.read_csv(r"C:\Users\Jairus\OneDrive\Documents\Portfolio\01.Fraud_Detection_Project\data\processed\y_test.csv").squeeze()

print(X_train.shape, X_test.shape, y_train.shape, y_test.shape)
```

# Model 1 — Logistic Regression

## Train logistic regression

```python
log_reg = LogisticRegression(max_iter=1000, random_state=42)
log_reg.fit(X_train, y_train)
```

## Logistic predictions

```python
y_pred_lr = log_reg.predict(X_test)
y_prob_lr = log_reg.predict_proba(X_test)[:, 1]
```

## Logistic evaluation

```python
print("Logistic Regression Classification Report")
print(classification_report(y_test, y_pred_lr))

print("Logistic Regression Confusion Matrix")
print(confusion_matrix(y_test, y_pred_lr))

print("Logistic Regression ROC-AUC:", roc_auc_score(y_test, y_prob_lr))
```

# Model 2 — Random Forest


## Train random forest

```python
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)
rf_model.fit(X_train, y_train)
```

## Random forest predictions

```python
y_pred_rf = rf_model.predict(X_test)
y_prob_rf = rf_model.predict_proba(X_test)[:, 1]
```

## Random forest evaluation

```python
print("Random Forest Classification Report")
print(classification_report(y_test, y_pred_rf))

print("Random Forest Confusion Matrix")
print(confusion_matrix(y_test, y_pred_rf))

print("Random Forest ROC-AUC:", roc_auc_score(y_test, y_prob_rf))
```

# Compare ROC-AUC

```python
print("Logistic Regression ROC-AUC:", roc_auc_score(y_test, y_prob_lr))
print("Random Forest ROC-AUC:", roc_auc_score(y_test, y_prob_rf))
```
